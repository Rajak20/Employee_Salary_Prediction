```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('employee_salary_regression.csv')
```

```python
df.drop("employee_id",axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['job_role'] = le.fit_transform(df['job_role'])
df['education_level'] = le.fit_transform(df['education_level'])
```

```python
df.head()
```

```python
x = df.drop("annual_salary_usd",axis=1)
y = df["annual_salary_usd"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_sc = scaler.fit_transform(X_train)
x_test_sc = scaler.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train_sc,y_train)
```

```python
y_pred = model.predict(x_test_sc)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
# Predict on the training set
y_train_pred = model.predict(x_train_sc)

# Calculate R2 score for both train and test sets
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_pred)

print(f"Training R2 Score: {r2_train:.4f}")
print(f"Testing R2 Score:  {r2_test:.4f}")

# Diagnosis
difference = r2_train - r2_test
if difference > 0.05:
    print(f"\nDiagnosis: Potential Overfitting. The train performance is better than test by {difference*100:.2f}%.")
else:
    print("\nDiagnosis: Good Model! The training and testing performances are very close, indicating great generalization.")
```

```python
import pickle

# 1. Save the model and the scaler
with open('salary_model.pkl', 'wb') as model_file:
    pickle.dump(model, model_file)

with open('scaler.pkl', 'wb') as scaler_file:
    pickle.dump(scaler, scaler_file)

print("Model and Scaler successfully saved using pickle!")
```
